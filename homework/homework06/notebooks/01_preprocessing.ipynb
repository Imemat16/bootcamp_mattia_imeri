```python
import os
import sys
import numpy as np
import pandas as pd

# Route paths so we can import from src/
sys.path.append(os.path.abspath('..'))
from src.cleaning import fill_missing_median, drop_missing, normalize_data

# Create messy dummy data
messy_data = pd.DataFrame({
    'id': [1, 2, 3, 4, 5, 6],
    'price': [100.0, 250.0, np.nan, 400.0, 500.0, 250.0],
    'category': ['A', 'B', 'A', np.nan, 'B', 'C'],
    'score': [10, 45, 30, 90, 100, 60]
})

# Save to data/raw/
raw_path = '../data/raw/messy_dataset.csv'
messy_data.to_csv(raw_path, index=False)
print("Raw messy data generated and saved!")
```

```python
# 1. Load Dataset
df_raw = pd.read_csv(raw_path)
print("--- ORIGINAL DATA ---")
display(df_raw)

# 2. Apply Cleaning Functions
# Fill missing prices with the median
df_clean = fill_missing_median(df_raw, column='price')

# Drop rows where 'category' is missing (since we can't mathematically impute text)
df_clean = drop_missing(df_clean, subset=['category'])

# Normalize the numeric 'price' and 'score' columns
df_clean = normalize_data(df_clean, columns=['price', 'score'])

print("\n--- CLEANED DATA ---")
display(df_clean)

# 3. Save processed dataset
processed_path = '../data/processed/cleaned_dataset.csv'
df_clean.to_csv(processed_path, index=False)
print(f"\nCleaned data saved to: {processed_path}")
```
